## Challenges and solutions

Since the data was collected by ourselves, the background of the images was chosen for the consistency. The first version of our dataset contained images with light background only. This raised the question whether the model could correctly predict a dish with black or colored background. We added second version of the dataset with more variation to see the impact of the background.

Because the images were taken with different devices, they varied in size and framing. The images were resized to 224x224 pixels to match the input size of the pre-trained model (VGG16). The compressed images are also less space-consuming in the github repository.

The images had to be divided into seperate sets for training, validation and testing. Since the data contains multiple images from the same dish, splitting the data automatically with script could lead to data leakage. If the same dish ended up in the training and testing sets, the model could predict the class correclty even though the features were never learned.
To prevent data leakage, we organized the images manually.

The slow training time limited the ability to re-run the model. The fine-tuned model was partically time consuming, so each change had to be planned carefully to reduce the number of unnecessary runs.
