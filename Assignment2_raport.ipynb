## Dataset description

The dataset for the model consists of images that have dinnerware items in them. There are images of four different types of dinnerware items: bowls, drinking glasses, mugs and plates. The dataset cointains 119 images of bowls, 120 images of drinking glasses, 120 images of mugs and 125 images of plates. The images were specifically taken for this assignment by all the team members meaning that the dataset was compiled with the model in mind. The dataset is still fairly small due to the limited time for searching and photographing different items. This can affect the performance of the model.

Most of the images have a light colored background but some were intentionally taken with different backgroud colours to hopefully improve the generalization ability of the model. All images show the target item on a simple background to not confuse the model too much. Most items are featured in at least two different images but from different angels. The items were intentionally photographed from different angels to improve the models training and generalization ability. The images are in RGB with 224x224 pixel dimensions.

The dataset is split in to training, validation and test sets. This wass done beforehand with the folder structure, meaning each subset has it own folder. Inside these folders are separate folders for each of the target classes. The split is aproximately 60-20-20, as in the training set has around 60% of the images and the validation and test sets have around 20% of the images. The training set contains 68 images of bowls, 74 images of drinking glasses, 78 images of mugs and 76 images of plates. The validation set contains 25 images of bowls, 22 images of drinking glasses, 21 images of mugs and 25 images of plates while the test set contains 26 images of bowls, 24 images of drinking glasses, 21 images of mugs and 24 images of plates

## Model performance comparison

| Model | Test accuracy (%) | Test loss | Training time (s) |
|----| ----|-----:| ----:|
| Own CNN | 72.6 | 0.672 | 486.8 |
| VGG16 feature extraction | 88.4 | 0.457 | 11.9 |
| Fine-tuned VGG16 | 93.7 | 0.970 | 923.4 |

Table 1. Results on the test set. The fine-tuned model's time is Phase 1 (395.4 s) plus Phase 2 (528.0 s). The feature extraction time covers only training the classifier. Computing the VGG16 feature took extra time that is not included. Times depend on the computer, so they are only indicative.

The CNN model trained from scratch classified 72.6% of the test images correctly. The feature extraction model, which uses a pretrained VGG16 base, did better and classified 88.4% of the images correctly. The fine-tuned model improved further and classified 93.7% correctly.

Fine-tuned model had the highest accuracy, but its test loss was higher than in the other two models. Loss punishes confident wrong predictions heavily, so even a few of them can raise the loss when accuracy is good. Fine-tuning still improved the model. Before fine-tuning (feature extraction with data augmentation) the test accuracy was 89.5% and the test loss 1.331. After fine-tuning they were 93.7% and loss 0.970.

The training time differs a lot between the models. CNN model took 487 s because it was trained for 70 epochs. Feature extraction model needed only 12 s because the VGG16 features were computed once and only the classifier was trained on them. The fine-tuned model was the slowest because it was trained in two phases. In both phases every epoch runs the augmented images through VGG16, so each epoch took longer than in the other two models.

## Challenges and solutions

Since the data was collected by ourselves, the background of the images was chosen for the consistency. The first version of our dataset contained images with light background only. This raised the question whether the model could correctly predict a dish with black or colored background. We added second version of the dataset with more variation to see the impact of the background.

Because the images were taken with different devices, they varied in size and framing. The images were resized to 224x224 pixels to match the input size of the pre-trained model (VGG16). The compressed images are also less space-consuming in the github repository.

The images had to be divided into seperate sets for training, validation and testing. Since the data contains multiple images from the same dish, splitting the data automatically with script could lead to data leakage. If the same dish ended up in the training and testing sets, the model could predict the class correclty even though the features were never learned.
To prevent data leakage, we organized the images manually.

The slow training time limited the ability to re-run the model. The fine-tuned model was partically time consuming, so each change had to be planned carefully to reduce the number of unnecessary runs.

## Suitability of the models

The CNN model does not seem to fit this dataset well, mainly because the dataset is small. Training a model from scratch needs a lot of data, and with only a few hundred images the model starts to overfit before it can reach a high accuracy. Adding more layers makes training slower without improving the results much. The CNN model had the weakest performance of the three models.

The feature extraction model uses the VGG16 convolutional base, which was trained on the ImageNet dataset to recognise general features, like edges, shapes and textures. Because our dataset is small, using these pretrained features is a large benefit, and the model performed clearly better than the CNN model. The features are extracted only once, so training is very fast. However, the VGG16 base isn't trained on our dataset at all, which limits the accuracy of the predictions.

The fine-tuned model also uses VGG16 but its last layers are trained further with our images. This made it the most accurate model. On the other hand, it is the largest model and took by far the longest to train. Its higher loss also means that some of its wrong predictions were made with high confidence.

Unless the goal is to be as accurate as possible, training time should also be considered when selecting the best model. If the model needs to be retrained as new images are added, the feature extraction model can be retrained much faster, with only slightly lower accuracy. Therefore, of the three models, feature extraction with VGG16 is the most suitable model for this use case.
