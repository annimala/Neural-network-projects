## Dataset description

The dataset for the model consists of images that have dinnerware items in them. There are images of four different types of dinnerware items: bowls, drinking glasses, mugs and plates. The dataset cointains 119 images of bowls, 120 images of drinking glasses, 120 images of mugs and 125 images of plates. The images were specifically taken for this assignment by all the team members meaning that the dataset was compiled with the model in mind. The dataset is still fairly small due to the limited time for searching and photographing different items. This can affect the performance of the model.

Most of the images have a light colored background but some were intentionally taken with different backgroud colours to hopefully improve the generalization ability of the model. All images show the target item on a simple background to not confuse the model too much. Most items are featured in at least two different images but from different angels. The items were intentionally photographed from different angels to improve the models training and generalization ability. The images are in RGB with 224x224 pixel dimensions.

The dataset is split in to training, validation and test sets. This wass done beforehand with the folder structure, meaning each subset has it own folder. Inside these folders are separate folders for each of the target classes. The split is aproximately 60-20-20, as in the training set has around 60% of the images and the validation and test sets have around 20% of the images. The training set contains 68 images of bowls, 74 images of drinking glasses, 78 images of mugs and 76 images of plates. The validation set contains 25 images of bowls, 22 images of drinking glasses, 21 images of mugs and 25 images of plates while the test set contains 26 images of bowls, 24 images of drinking glasses, 21 images of mugs and 24 images of plates

## Challenges and solutions

Since the data was collected by ourselves, the background of the images was chosen for the consistency. The first version of our dataset contained images with light background only. This raised the question whether the model could correctly predict a dish with black or colored background. We added second version of the dataset with more variation to see the impact of the background.

Because the images were taken with different devices, they varied in size and framing. The images were resized to 224x224 pixels to match the input size of the pre-trained model (VGG16). The compressed images are also less space-consuming in the github repository.

The images had to be divided into seperate sets for training, validation and testing. Since the data contains multiple images from the same dish, splitting the data automatically with script could lead to data leakage. If the same dish ended up in the training and testing sets, the model could predict the class correclty even though the features were never learned.
To prevent data leakage, we organized the images manually.

The slow training time limited the ability to re-run the model. The fine-tuned model was partically time consuming, so each change had to be planned carefully to reduce the number of unnecessary runs.
