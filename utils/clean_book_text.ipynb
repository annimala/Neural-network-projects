```python
import re

# File names
input_file = "../datasets/jane_austen_pride_and_prejudice.txt"
output_file = "../datasets/jane_austen_pride_and_prejudice_clean.txt"

# Line numbers (as shown in a text editor)
remove_until = 692      # removes lines 1 to x
remove_from = 14549    # removes line y to the end

def clean_text(text):
    # Remove [ ] and everything inside (repeat to handle nested brackets)
    previous = None
    while previous != text:
        previous = text
        text = re.sub(r"\[[^\[\]]*\]", "", text)

    # Remove /* */ markers and layout tags (NIND, RIGHT), keep the text inside
    text = re.sub(r"/\*\s*(?:[A-Z]{3,}\s+)?(.*?)\s*\*/", r"\1", text, flags=re.DOTALL)

    # Remove all underscores
    text = text.replace("_", "")

    # Tidy up leftover double spaces and trailing spaces
    text = re.sub(r"[ \t]{2,}", " ", text)
    text = "\n".join(line.strip() for line in text.splitlines())

    # Join wrapped lines: one line per paragraph, blank line between paragraphs
    paragraphs = re.split(r"\n\s*\n", text)
    paragraphs = [" ".join(p.split("\n")).strip() for p in paragraphs]
    text = "\n\n".join(p for p in paragraphs if p) + "\n"

    # Restore the first chapter heading (it was inside an illustration bracket)
    text = "CHAPTER I.\n\n" + text
    return text

with open(input_file, "r", encoding="utf-8") as f:
    lines = f.read().splitlines()

lines = lines[remove_until : remove_from - 1]
text = "\n".join(lines)

cleaned = clean_text(text)

with open(output_file, "w", encoding="utf-8") as f:
    f.write(cleaned)

print(f"Done! Saved to {output_file}")
```
