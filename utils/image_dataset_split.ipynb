```python
import random
import shutil
from pathlib import Path

FOLDERS = [
    Path("../dinnerware/bowl"),
    Path("../dinnerware/glass"),
    Path("../dinnerware/mug"),
    Path("../dinnerware/plate"),
]
new_base_dir = Path("../dinnerware")

TRAIN_RATIO = 0.7
VALIDATION_RATIO = 0.15      # test gets the rest

def make_subset(subset_name, folder, fnames):
    dir = new_base_dir / subset_name / folder.name
    dir.mkdir(parents=True)                  # stops with an error if the folder already exists
    for fname in fnames:
        shutil.copyfile(src=folder / fname, dst=dir / fname)

for subset_name in ("train", "validation", "test"):
    subset_dir = new_base_dir / subset_name
    if subset_dir.exists():
        shutil.rmtree(subset_dir)    # only this split folder; anything else in new_base_dir stays

for folder in FOLDERS:
    fnames = sorted(p.name for p in folder.iterdir() if p.is_file())
    random.Random(0).shuffle(fnames)         # same shuffle every run

    n = len(fnames)
    train_end = int(n * TRAIN_RATIO)
    validation_end = train_end + int(n * VALIDATION_RATIO)

    make_subset("train", folder, fnames[:train_end])
    make_subset("validation", folder, fnames[train_end:validation_end])
    make_subset("test", folder, fnames[validation_end:])

    print(f"{folder.name}: {n} total, train {train_end}, "
          f"validation {validation_end - train_end}, test {n - validation_end}")
```
